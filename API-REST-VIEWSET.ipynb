# src\rest_examples\views.py:

```python
from rest_framework.views  import APIView
from rest_framework.response  import Response
from rest_framework import status

from rest_framework import viewsets
from rest_framework.viewsets import ViewSet

from products.models import Product #<----
from .serializers import ProductSerializer, TestSerializers #<----

# Create your views here.
     
class ProductViewSet(viewsets.ModelViewSet): #<----

    queryset = Product.objects.all()
    serializer_class = ProductSerializer

class TestAPIView(APIView):
    "API View de Prueba " 
    serializer_class = TestSerializers
     
    def get(self, request, format=None):
        """Regresa una lista de caractristicas de un APIView"""
        apiview_info = [
            "Usa metodos HTTP como funciones (get, post, patch, put, delete)",
            "Es similar a un Djanfo View tadicional",
            "Te da el mayor control de la logica de la app",
            "Es mapeado manualmente a los urls",   
        ]
         
        return Response({"message":"Hola", "apiview_info":apiview_info})
    
    def post(self, request):
        """Crear un mensaje con el nombre ingresado"""
        serializer = self.serializer_class(data=request.data)
        
        if serializer.is_valid():
            name = serializer.validated_data.get("name")
            message = f"Hola {name}"
            return Response({"message": message})
        else:
            return Response(serializer.errors, status=status.HTTP_400_BAD_REQUEST)
        
    def put(self, request, pk=None):
            """Manejar la actualizacion de un objeto"""
            return Response({"method": "PUT"})

    def patch(self, request, pk=None):
            """Manejar la actualizacion de un objeto parcialmente"""
            return Response({"method": "PATCH"})
       
    def delete(self, request, pk=None):
            """Manejar la eliminacion de un objeto"""
            return Response({"method": "DELETE"})
       
                             
    def partial_update(self, request, pk=None):
        """Manejar la consulta parcial de un objeto por su id"""
        return Response({"method": "PACTH"})
    
    def destroy(self, request, pk=None):
        """Manejar la eliminacion de un objeto por su id"""
        return Response({"method": "DELETE"})
    
    
class TestViewSet(ViewSet) :
    """Test ViewSet"""
    serializer_class = TestSerializers

    
    def list(self, request):
        """"Regresa un listado de caracteristicas de un ViewSet"""                          
        viewsets = [
            "Usa acciones (list, create, retrive, update, partial_update, destroy)",
            "Se mapea automaticamente a los urls usando Routers",
            "Provee mas funcionalidad con menos codigo",  
        ]
        return Response({"message": "Hola", "viewsets_info": viewsets})
    
    def create(self, request):
        """Crear un mensaje de saludo"""
        serializer = self.serializer_class(data=request.data)
        
        if serializer.is_valid():
            name = serializer.validated_data.get("name")
            message = f"Hola {name}"
            return Response({"message": message})
        else:
            return Response(serializer.errors, status=status.HTTP_400_BAD_REQUEST)
        
    def retrive(self, request, pk=None):
        """Manejar la consulta de un objeto por su id"""
        return Response({"method": "GET"})
    
    def update(self, request, pk=None):
        """Manejar la actualizacion de un objeto por su id"""
        return Response({"method": "PUT"})
```
