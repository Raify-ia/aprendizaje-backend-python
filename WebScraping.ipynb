# Código completo en Jupyter Notebook detalladamente

```python
from selenium import webdriver
from bs4 import BeautifulSoup
import time

driver = webdriver.Chrome()

url = "https://www.amazon.com/s?k=laptop"

driver.get(url)

time.sleep(5)

html = driver.page_source

soup = BeautifulSoup(html, "html.parser")

[data-component-type="s-search-result"]

products = soup.select('[data-component-type="s-search-result"]')

len(products)

links = []

for product in products:
    
    a_tag = product.select_one("a.a-link-normal")
    
    if a_tag:
        
        href = a_tag.get("href")
        
        if href:
            full_link = "https://www.amazon.com" + href
            
            links.append(full_link)

links[:5]

product_url = links[0]

driver.get(product_url)

time.sleep(5)

detail_html = driver.page_source

detail_soup = BeautifulSoup(detail_html, "html.parser")

title = detail_soup.select_one("#productTitle")

print(title.text.strip())

price = detail_soup.select_one(".a-price .a-offscreen")

print(price.text.strip())

description = detail_soup.select_one("#feature-bullets")

print(description.text.strip())
```
